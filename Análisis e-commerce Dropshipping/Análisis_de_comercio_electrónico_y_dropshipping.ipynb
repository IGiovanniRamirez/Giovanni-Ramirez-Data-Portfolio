## 📌 1. Contexto de negocio

Este proyecto se basa en datos reales de una operación de e-commerce bajo el modelo de **dropshipping con pago contra entrega (COD)** en Colombia.

El negocio operaba a través de:

- **Shopify** → canal comercial (tienda online)  
- **Dropi** → operación logística y financiera  
- **Meta Ads (Facebook / Instagram) y TikTok Ads** → adquisición de clientes  

A diferencia de análisis tradicionales, este proyecto incorpora una visión **end-to-end**, integrando:

- Ingresos  
- Costos de producto  
- Costos logísticos  
- Costos de marketing  

Esto permite analizar la **rentabilidad real del negocio**, más allá de métricas superficiales.

---

### ⚠️ Desafíos del modelo COD

- Alto riesgo de pedidos no efectivos (cancelaciones y devoluciones)  
- Costos logísticos elevados (envíos fallidos y devoluciones)  
- Dependencia de validación manual de pedidos  
- El marketing se paga incluso si el pedido no se entrega  

---

## 🔁 2. Proceso operativo del negocio

Flujo real de la operación:

1. Cliente ve anuncio (Meta Ads / TikTok Ads)  
2. Cliente realiza pedido en Shopify  
3. Confirmación automática vía chatbot (Dropi)  
4. Validación manual si no hay respuesta, cancelación de pedido en caso de tampoco encontrar respuesta en la validación manual.  
5. Procesamiento logístico  
6. Resultado del pedido:

- ✅ **ENTREGADO** → venta exitosa  
- ⚠️ **DEVOLUCIÓN / DEVUELTA** → pérdida logística  
- ❌ **CANCELADO / RECHAZADO** → pedido no ejecutado  
- ⚫ **RECOGIDA FALLIDA** → pedido no procesado  
- 🔴 **INDEMNIZADA POR DROPI** → pérdida total (producto + envío + reembolso)  

---

- Pérdidas operativas

# Data Cleaning (Python - Pandas) 👇

```python
# =========================
# 1. LIBRERÍAS
# =========================
import pandas as pd
import numpy as np

# =========================
# 2. COPIA DE SEGURIDAD
# =========================
orders = orders_products_raw.copy()
customers = customers_raw.copy()
marketing = marketing_raw.copy()

# =========================
# 3. LIMPIEZA: orders_products_raw
# =========================

# --- 3.1 Eliminar columnas irrelevantes o vacías ---
cols_drop_orders = [
    'TIPO DE IDENTIFICACION',
    'NRO DE IDENTIFICACION',
    'OBSERVACIÓN',
    'TAGS',
    'CONTADOR DE INDEMNIZACIONES',
    'CONCEPTO ÚLTIMA INDENMIZACIÓN',
    'VENDEDOR',
    'TIPO DE TIENDA',
    'TIENDA',
    'ID DE ORDEN DE TIENDA',
    'NUMERO DE PEDIDO DE TIENDA'
]

orders.drop(columns=cols_drop_orders, inplace=True, errors='ignore')

# --- 3.2 Conversión de fechas ---
date_cols = [
    'FECHA',
    'FECHA GUIA GENERADA',
    'FECHA DE NOVEDAD',
    'FECHA DE SOLUCIÓN',
    'FECHA DE ÚLTIMO MOVIMIENTO'
]

for col in date_cols:
    orders[col] = pd.to_datetime(orders[col], errors='coerce', dayfirst=True)

# --- 3.3 Limpieza de texto ---
text_cols = [
    'NOMBRE CLIENTE',
    'CIUDAD DESTINO',
    'DEPARTAMENTO DESTINO',
    'PRODUCTO',
    'VARIACION'
]

for col in text_cols:
    orders[col] = (
        orders[col]
        .astype(str)
        .str.strip()
        .str.upper()
    )

# --- 3.4 Teléfono como string ---
orders['TELÉFONO'] = orders['TELÉFONO'].astype(str).str.replace('.0', '', regex=False)

# --- 3.5 Normalización de ESTATUS ---
def map_status(status):
    if status in ['ENTREGADO']:
        return 'delivered'
    elif status in ['DEVOLUCION', 'DEVUELTA']:
        return 'returned'
    elif status in ['CANCELADO', 'RECHAZADO']:
        return 'cancelled'
    elif status in ['RECOGIDA FALLIDA']:
        return 'not_processed'
    elif status in ['INDEMNIZADA POR DROPI']:
        return 'critical_loss'
    else:
        return 'other'

orders['status_group'] = orders['ESTATUS'].apply(map_status)

# --- 3.6 Eliminar duplicados exactos (no por ID solamente) ---
orders.drop_duplicates(subset=['ID', 'PRODUCTO', 'VARIACION'], inplace=True)

# --- 3.7 Feature engineering (renombrado de columnas) ---

orders.rename(columns={
    'TOTAL DE LA ORDEN': 'revenue',
    'PRECIO PROVEEDOR X CANTIDAD': 'cost_product',
    'PRECIO FLETE': 'cost_shipping',
    'COSTO DEVOLUCION FLETE': 'cost_return'
}, inplace=True)

# estandarizar a snake_case
orders.columns = (
    orders.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('á', 'a')
    .str.replace('é', 'e')
    .str.replace('í', 'i')
    .str.replace('ó', 'o')
    .str.replace('ú', 'u')
    .str.replace('ñ', 'n')
)

# Flags
orders['is_delivered'] = (orders['status_group'] == 'delivered').astype(int)
orders['is_returned'] = (orders['status_group'] == 'returned').astype(int)
orders['is_cancelled'] = (orders['status_group'] == 'cancelled').astype(int)

# Definición de pérdida
orders['is_loss'] = np.where(
    orders['status_group'].isin(['returned', 'critical_loss']),
    1, 0
)

# --- 3.8 Validaciones básicas ---
# Reemplazar valores negativos por NaN (si existen)
num_cols = ['revenue', 'cost_product', 'cost_shipping', 'cost_return']

for col in num_cols:
    orders[col] = orders[col].apply(lambda x: np.nan if x < 0 else x)

# =========================
# 4. LIMPIEZA: customers_raw
# =========================

# --- 4.1 Eliminar columnas vacías ---
customers.drop(columns=['ADDRESS2', 'IDENTIFICATION'], inplace=True, errors='ignore')

# --- 4.2 Limpieza de nombres ---
customers['NAME'] = customers['NAME'].astype(str).str.strip().str.title()
customers['LASTNAME'] = customers['LASTNAME'].astype(str).str.strip().str.title()

# --- 4.3 Crear full_name ---
customers['full_name'] = customers['NAME'] + ' ' + customers['LASTNAME']

# --- 4.4 Teléfono como string ---
customers['PHONE'] = customers['PHONE'].astype(str).str.replace('.0', '', regex=False)

# --- 4.5 Limpieza de dirección ---
customers['ADDRESS1'] = customers['ADDRESS1'].astype(str).str.strip().str.upper()

# estandarizar a snake_case
customers.columns = (
    customers.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('á', 'a')
    .str.replace('é', 'e')
    .str.replace('í', 'i')
    .str.replace('ó', 'o')
    .str.replace('ú', 'u')
    .str.replace('ñ', 'n')
)

# =========================
# 5. LIMPIEZA: marketing_raw
# =========================

# --- 5.1 Estandarizar nombres ---

marketing.columns = (
    marketing.columns
    .str.strip()
    .str.lower()
    .str.replace('ñ', 'n')
)
marketing.rename(columns={
    'canal': 'channel',
    'costo': 'marketing_cost'
    }, inplace=True)

marketing['channel'] = marketing['channel'].replace({
    'Meta ads': 'meta_ads',
    'Tiktok ads': 'tiktok_ads',
    'Shopify': 'shopify'
})

# --- 5.2 Crear número de mes ---
month_map = {
    'enero': 1, 'febrero': 2, 'marzo': 3, 'abril': 4,
    'mayo': 5, 'junio': 6, 'julio': 7, 'agosto': 8,
    'septiembre': 9, 'octubre': 10, 'noviembre': 11, 'diciembre': 12
}
marketing['mes'] = (marketing['mes'].astype(str).str.strip().str.lower())
marketing['month_number'] = marketing['mes'].map(month_map)

# --- 5.3 Crear fecha ---
marketing['date'] = pd.to_datetime(
    dict(year=marketing['ano'], month=marketing['month_number'], day=1)
)
marketing['marketing_cost'] = pd.to_numeric(marketing['marketing_cost'], errors='coerce')

# =========================
# 6. OUTPUT LIMPIO
# =========================

orders_clean = orders.copy()
customers_clean = customers.copy()
marketing_clean = marketing.copy()

# Vista rápida
print("Orders limpio:", orders_clean.shape)
print("Customers limpio:", customers_clean.shape)
print("Marketing limpio:", marketing_clean.shape)
```

# Data Integration (tabla final unificada) 👇

## 🧱 2. Resultado

Se construye la tabla:  `fact_final`

Esta tabla contiene:

- Información transaccional integrada  
- Costos operativos ajustados  
- Costo de marketing prorrateado  


### Crear tabla `fact_final`

```python
conn.execute("DROP TABLE IF EXISTS fact_final")

query = """
CREATE TABLE fact_final AS
WITH marketing_monthly AS (
    SELECT
        STRFTIME('%Y', date) AS year,
        STRFTIME('%m', date) AS month,
        SUM(marketing_cost) AS total_marketing_cost
    FROM marketing_clean
    GROUP BY year, month
),
orders_per_month AS (
    SELECT
        STRFTIME('%Y', fecha) AS year,
        STRFTIME('%m', fecha) AS month,
        COUNT(DISTINCT id) AS total_orders
    FROM orders_clean
    GROUP BY year, month
)

SELECT
    -- Identificadores
    o.id AS order_id,
    DATE(o.fecha) AS date,

    -- Cliente
    c.id AS customer_id,

    -- Producto
    o.producto AS product_name,
    o.variacion AS variation,
    o.cantidad AS quantity,

    -- Ubicación
    o.departamento_destino AS department,
    o.ciudad_destino AS city,

    -- Tiempo
    CAST(STRFTIME('%Y', o.fecha) AS INTEGER) AS year,
    CAST(STRFTIME('%m', o.fecha) AS INTEGER) AS month,

    -- Estado
    o.status_group,

    -- Revenue real
    CASE
        WHEN o.status_group = 'delivered' THEN o.revenue
        ELSE 0
    END AS revenue_real,

    -- Costos reales
    CASE
        WHEN o.status_group IN ('delivered','returned','critical_loss')
        THEN o.cost_product ELSE 0
    END AS cost_product_real,

    CASE
        WHEN o.status_group IN ('delivered','returned','critical_loss')
        THEN o.cost_shipping ELSE 0
    END AS cost_shipping_real,

    CASE
        WHEN o.status_group = 'returned'
        THEN o.cost_return ELSE 0
    END AS cost_return_real,

    -- Marketing prorrateado (unitario)
    (m.total_marketing_cost * 1.0 / opm.total_orders) AS marketing_cost_per_order,

       -- Flags
    o.is_delivered,
    o.is_returned,
    o.is_cancelled,
    o.is_loss

FROM orders_clean o

LEFT JOIN customers_clean c
    ON o.telefono = c.phone

LEFT JOIN marketing_monthly m
    ON STRFTIME('%Y', o.fecha) = m.year
    AND STRFTIME('%m', o.fecha) = m.month

LEFT JOIN orders_per_month opm
    ON STRFTIME('%Y', o.fecha) = opm.year
    AND STRFTIME('%m', o.fecha) = opm.month;
"""
conn.execute(query)
```

### Validación

# KPI Calculation (Operativos + Marketing) 👇

## 📌 1. Enfoque del análisis
En esta etapa se calculan los indicadores clave de desempeño (KPIs) a partir de la tabla integrada `fact_final`.

A diferencia del EDA, aquí se construyen métricas agregadas que permiten evaluar de forma clara:

- La rentabilidad del negocio

Todos los KPIs se calculan utilizando consultas SQL sobre datos ya integrados y limpios.

## 🎯 2. Objetivos de los KPIs

- Medir la rentabilidad del negocio (operativa y neta)
- Evaluar el desempeño logístico
- Cuantificar el impacto de devoluciones y cancelaciones
- Incorporar el costo de marketing en el análisis

---

## 💰 3. KPIs Financieros

Se calculan métricas agregadas a nivel global:

### Ingresos y costos
- **Revenue Total** → suma de `revenue`
- **Costo de producto (COGS)** → suma de `cost_product`

- **Costo de marketing total** → suma de `marketing_cost`

### 📈 Rentabilidad

- **Profit Operativo**
  
  `revenue - (cost_product + cost_shipping + cost_return)`

- **Profit Neto**
  
  `profit_operativo - marketing_cost`

- **Margen Operativo**
  
  `profit_operativo / revenue`

- **Margen Neto**
  
  `profit_neto / revenue`

---

## 🚚 4. KPIs Operativos

- **Fulfillment Rate**
  
  `% pedidos entregados` → `is_delivered`

- **Cancellation Rate**
  
  `% pedidos cancelados` → `is_cancelled`

- **Non-Processed Rate**
  
  `% pedidos no procesados` → `status_group = 'not_processed'`

## ⚠️ 5. KPIs de Riesgo

- **Loss Rate**
  
  `% pedidos con pérdida` → `is_loss`

  `% pedidos con pérdida crítica` → `status_group = 'critical_loss'`

## 📣 6. KPIs de Marketing

  suma de `marketing_cost`

  `marketing_cost / total_orders`

### Base a nivel pedido

```python
query = """
WITH order_level AS (
    SELECT
        order_id,

        SUM(revenue_real) AS revenue,
        SUM(cost_product_real) AS cost_product,
        SUM(cost_shipping_real) AS cost_shipping,
        SUM(cost_return_real) AS cost_return,

        MAX(marketing_cost_per_order) AS marketing_cost,

        MAX(is_delivered) AS is_delivered,
        MAX(is_returned) AS is_returned,
        MAX(is_cancelled) AS is_cancelled,
        MAX(is_loss) AS is_loss,

        MAX(CASE WHEN status_group = 'critical_loss' THEN 1 ELSE 0 END) AS is_critical_loss

    FROM fact_final
    GROUP BY order_id
)
SELECT * FROM order_level LIMIT 10;
"""
run_query(query)
```

### KPIs financieros principales

```python
query = """
WITH order_level AS (
    SELECT
        order_id,
        SUM(revenue_real) AS revenue,
        SUM(cost_product_real) AS cost_product,
        SUM(cost_shipping_real) AS cost_shipping,
        SUM(cost_return_real) AS cost_return,
        MAX(marketing_cost_per_order) AS marketing_cost
    FROM fact_final
    GROUP BY order_id
)

SELECT
    SUM(revenue) AS total_revenue,
    SUM(cost_product) AS total_cogs,
    SUM(cost_shipping) AS total_shipping,
    SUM(cost_return) AS total_return_cost,
    SUM(marketing_cost) AS total_marketing,

    -- Profit operativo
    SUM(revenue - cost_product - cost_shipping - cost_return) AS profit_operativo,

    -- Profit neto
    SUM(revenue - cost_product - cost_shipping - cost_return - marketing_cost) AS profit_neto

FROM order_level;
"""
run_query(query)
```

### Márgenes

```python
query = """
WITH order_level AS (
    SELECT
        order_id,
        SUM(revenue_real) AS revenue,
        SUM(cost_product_real) AS cost_product,
        SUM(cost_shipping_real) AS cost_shipping,
        SUM(cost_return_real) AS cost_return,
        MAX(marketing_cost_per_order) AS marketing_cost
    FROM fact_final
    GROUP BY order_id
)

SELECT
    SUM(revenue) AS total_revenue,

    -- Margen operativo
    ROUND(
        SUM(revenue - cost_product - cost_shipping - cost_return) * 1.0
        / NULLIF(SUM(revenue),0), 4
    ) AS margin_operativo,

    -- Margen neto
    ROUND(
        SUM(revenue - cost_product - cost_shipping - cost_return - marketing_cost) * 1.0
        / NULLIF(SUM(revenue),0), 4
    ) AS margin_neto

FROM order_level;
"""
run_query(query)
```

### KPIs operativos

```python
query = """
WITH order_level AS (
    SELECT
        order_id,
        MAX(is_delivered) AS is_delivered,
        MAX(is_returned) AS is_returned,
        MAX(is_cancelled) AS is_cancelled,
        MAX(CASE WHEN status_group = 'not_processed' THEN 1 ELSE 0 END) AS is_not_processed
    FROM fact_final
    GROUP BY order_id
)

SELECT
    COUNT(*) AS total_orders,

    ROUND(SUM(is_delivered) * 1.0 / COUNT(*), 4) AS fulfillment_rate,
    ROUND(SUM(is_returned) * 1.0 / COUNT(*), 4) AS return_rate,
    ROUND(SUM(is_cancelled) * 1.0 / COUNT(*), 4) AS cancellation_rate,
    ROUND(SUM(is_not_processed) * 1.0 / COUNT(*), 4) AS not_processed_rate

FROM order_level;
"""
run_query(query)
```

### KPIs de riesgo

```python
query = """
WITH order_level AS (
    SELECT
        order_id,
        MAX(is_loss) AS is_loss,
        MAX(CASE WHEN status_group = 'critical_loss' THEN 1 ELSE 0 END) AS is_critical_loss
    FROM fact_final
    GROUP BY order_id
)

SELECT
    COUNT(*) AS total_orders,

    ROUND(SUM(is_loss) * 1.0 / COUNT(*), 4) AS loss_rate,
    ROUND(SUM(is_critical_loss) * 1.0 / COUNT(*), 4) AS critical_loss_rate

FROM order_level;
"""
run_query(query)
```

### KPIs de marketing

```python
query = """
WITH order_level AS (
    SELECT
        order_id,
        MAX(marketing_cost_per_order) AS marketing_cost,
        MAX(is_delivered) AS is_delivered
    FROM fact_final
    GROUP BY order_id
)

SELECT
    SUM(marketing_cost) AS total_marketing_cost,

    AVG(marketing_cost) AS marketing_cost_per_order,

    SUM(marketing_cost) * 1.0 / NULLIF(SUM(is_delivered),0)
        AS marketing_cost_per_delivered_order

FROM order_level;
"""
run_query(query)
```

### Profit por tipo de pedido

```python
query = """
WITH order_level AS (
    SELECT
        order_id,
        MAX(status_group) AS status_group,

        SUM(revenue_real) AS revenue,
        SUM(cost_product_real) AS cost_product,
        SUM(cost_shipping_real) AS cost_shipping,
        SUM(cost_return_real) AS cost_return,
        MAX(marketing_cost_per_order) AS marketing_cost
    FROM fact_final
    GROUP BY order_id
)

SELECT
    status_group,
    COUNT(*) AS total_orders,

    SUM(revenue - cost_product - cost_shipping - cost_return) AS profit_operativo,
    SUM(revenue - cost_product - cost_shipping - cost_return - marketing_cost) AS profit_neto

FROM order_level
GROUP BY status_group
ORDER BY profit_neto DESC;
"""
run_query(query)
```

# Insights & Business Recommendations 👇


## 🟦 1. Insight Principal: Rentabilidad Baja pero Positiva

El negocio presenta una rentabilidad neta limitada:

- **Revenue:** $47,195,297  

- **Profit Neto:** $928,281  

- **Margen Neto:** 2.0%

👉 **Conclusión:**

> ⚠️ El negocio es rentable, pero con un margen muy bajo, altamente sensible a cambios en la operación o en marketing.

---

## 🟥 2. Problema Crítico: Baja Efectividad Operativa

Distribución de pedidos:

- **Delivered:** 57.8%  
- **Cancelled:** 34.7%  
- **Returned:** 6.8%  

> ❗ El **42% de los pedidos no generan ingresos**, pero sí generan costos.


👉 A primera vista:

> ❗ Alto nivel de cancelaciones podría parecer ineficiencia

👉 Pero en realidad:

> 🧠 Las cancelaciones son parte de una **estrategia operativa deliberada** para evitar pérdidas mayores por devoluciones.

---

## 🔁 3. Rol Estratégico de las Cancelaciones

En el modelo COD:

- Un pedido **cancelado**:
  - ❌ No genera revenue
  - ❌ No incurre en costos logísticos
  - ✔ Solo pierde el costo de marketing

- Un pedido **devuelto**:
  - ❌ No genera revenue
  - ❌ Sí incurre en costos logísticos (envío + retorno)
  - ❌ Genera pérdida significativamente mayor

---

### 💣 Impacto financiero por tipo de pedido:

- Pedido entregado → **+ $29,576**

- Pedido cancelado → **- $21,195**

- Pedido devuelto → **- $94,107**

- Pedido con pérdida crítica → **- $98,748**

👉 **Insight clave:**

> 🔴 Una devolución puede costar hasta **4–5 veces más que una cancelación**

---

## 🧠 4. Insight Estratégico Principal

> 🔥 El negocio está optimizando pérdidas priorizando cancelaciones sobre devoluciones

👉 **Interpretación:**

- Las cancelaciones **no son completamente negativas**

- Funcionan como un **mecanismo de control de riesgo**

---

## ⚠️ 5. Problema Real: Calidad de la Demanda (Marketing + Oferta)

Aunque la estrategia de cancelación reduce pérdidas logísticas:

> ❗ Existe un problema más profundo: **la calidad de los pedidos generados**

---

### Evidencia:

- Alta proporción de pedidos cancelados (34.7%)
- Bajo margen neto (2%)
- Profit por pedido muy bajo ($1,439)

---

👉 **Insight clave:**

> ⚠️ El negocio está atrayendo una gran cantidad de clientes con baja intención real de compra

---

## 📣 6. Marketing: Genera volumen, pero no asegura calidad

- **ROAS:** 326%  

- **Marketing Cost:** $14.478.278  

👉 **Interpretación inicial:**
El marketing parece altamente rentable.

👉 **Pero en realidad:**

> ⚠️ El alto ROAS no se traduce en rentabilidad neta.

### ¿Por qué ocurre esto?

- El marketing genera pedidos, pero no controla su calidad
- Alta tasa de cancelaciones y fallos operativos
- Costos logísticos reducen significativamente el margen

👉 **Conclusión:**

> El marketing es eficiente en adquisición, pero ineficiente en generación de valor real.

---

## 🧠 7. Problema Estructural: Desalineación Oferta–Producto–Cliente

El problema no está solo en la operación, sino en la **propuesta de valor completa**:

---

### 🔍 Posibles causas:

#### 1. Oferta poco atractiva
- No genera urgencia o confianza suficiente

#### 2. Producto mal alineado con la necesidad
- No resuelve claramente un problema real

#### 3. Ángulo de venta débil o incorrecto
- No conecta con el público objetivo

#### 4. Expectativas mal comunicadas
- Diferencia entre lo que promete el anuncio y lo que recibe el cliente

---

👉 **Insight clave:**

> ❗ El marketing está trayendo volumen, pero no clientes de calidad

---

Por pedido:

> ⚠️ El margen es demasiado bajo para sostener errores operativos o mala calidad de demanda

## 🚚 9. Logística: Importante, pero no el problema principal

- **Shipping Cost:** $6,997,637  

- **Return Cost:** $976,411

> La logística impacta la rentabilidad, pero el negocio ya está intentando controlarla mediante cancelaciones

---

👉 **Conclusión:**

> 🔸 La logística es un costo relevante  
> 🔸 Pero el problema raíz está en la generación de pedidos (demanda)

## 📉 10. Escalabilidad Riesgosa

👉 Aumentar volumen sin mejorar calidad implica:

- Más cancelaciones  
- Más devoluciones  
- Mayor presión sobre márgenes  

---

> ⚠️ Escalar marketing sin optimizar la conversión real puede destruir valor

## 🚀 11. Recomendaciones Estratégicas

### 🎯 1. Mejorar calidad de la demanda

- Refinar segmentación
- Ajustar campañas
- Optimizar targeting

### 🎯 2. Rediseñar la oferta

- Mayor claridad en beneficios
- Incentivos más fuertes
- Mejor propuesta de valor

- Validar que el producto resuelve una necesidad real
- Ajustar portafolio si es necesario

### 🎯 4. Optimizar el ángulo de venta

- Mejor storytelling
- Comunicación clara
- Expectativas realistas

### 🎯 5. Mantener estrategia de cancelación (pero optimizarla)

- Reducir cancelaciones mejorando calidad de leads
