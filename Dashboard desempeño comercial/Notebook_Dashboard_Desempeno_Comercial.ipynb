## 🔹 Paso 2: Transformación con Power Query

---

En esta etapa se realizaron ajustes estructurales en Power Query con el objetivo de garantizar calidad, consistencia y preparación óptima del dataset para el análisis comercial.

## 1️⃣ Cambio de formato de fecha a Español (Latinoamérica)

La columna Fecha_Pedido fue configurada con formato regional Español (Latinoamérica).

**¿Por qué es importante?**

- Asegura correcta interpretación de fechas (día/mes/año).
- Evita errores en segmentaciones temporales.
- Permite análisis coherente de meses, estaciones y evolución anual.
- Garantiza consistencia con el contexto geográfico del negocio (Perú, Chile y Colombia).

**📌 Resultado:** La columna quedó correctamente estructurada como tipo Fecha, lista para análisis temporal y creación de jerarquías (Día → Mes → Año).

## 2️⃣ Corrección de tipos de datos numéricos

Se validaron y corrigieron los tipos de datos de las columnas financieras y cuantitativas:

- Unidades_Vendidas → Número entero
- Precio_Unitario → Número entero
- Ingresos → Número entero
- Costo → Número decimal

**¿Por qué es clave esta validación?**

- Evita errores en cálculos de margen y rentabilidad.
- Permite agregaciones correctas (SUM, AVERAGE, etc.).
- Previene que Power BI interprete números como texto.
- Asegura precisión en métricas financieras.

**📌 Resultado:** Todas las columnas cuantitativas quedaron correctamente tipadas para cálculos financieros confiables.

## 3️⃣ Creación de columna condicional: Nivel_Venta

Se creó una nueva columna condicional llamada Nivel_Venta, basada en la columna Ingresos.

**Reglas aplicadas:**

- Si Ingresos ≥ 1000 → "Venta Alta"
- De lo contrario → "Venta Baja"

**Objetivo estratégico**

Esta variable permitirá:

- Analizar proporción de ventas de alto valor.
- Identificar segmentos o categorías con mayor concentración de ventas relevantes.
- Detectar posibles oportunidades para incrementar ticket promedio.
- Facilitar análisis comparativos entre mercados.

**📌 Resultado:** El dataset ahora incluye una variable categórica adicional que permite segmentación estratégica por nivel de venta.

## 4️⃣ Validación con Vista Perfil de Columna

Se activó la opción Vista → Perfil de columna en Power Query para validar:

- Porcentaje de valores válidos.
- Valores vacíos o nulos.
- Distribución de datos.

- Posibles inconsistencias.

**Hallazgos**

## 🔹 Paso 5: Narrativa con Modelo SCQA + mensaje asincrónico

---

En esta sección se construye la narrativa estratégica del dashboard utilizando el modelo **SCQA (Situación, Complicación, Pregunta, Respuesta)**.  
El objetivo es traducir las visualizaciones en **insights claros que ayuden a la toma de decisiones del negocio**.

## Narrativa del Dashboard (Modelo SCQA)

## 🖥️ Vista General (Overview)

### S (Situación)
Andes Retail Group opera en los mercados de **Perú, Chile y Colombia**, comercializando productos en las categorías **Electrónica, Ropa, Deportes y Hogar**.  

El análisis del desempeño comercial entre **2024 y 2025** muestra:

- **Ingresos totales:** $5.53 millones
  
- **Margen bruto:** $1.94 millones

- **Margen promedio:** 35.10 %

- **Unidades vendidas:** 57.6 mil

Estos indicadores reflejan el desempeño consolidado del negocio en los dos años analizados.

### C (Complicación)
Al comparar el desempeño entre años se observa un **crecimiento significativo del negocio**:

- **Ingresos:** +93.2 %  
- **Margen bruto:** +93.1 %  
- **Unidades vendidas:** +95.7 %  

Este crecimiento indica una **expansión importante en el volumen de ventas**, aunque el **porcentaje de margen se mantiene relativamente estable**, lo que sugiere que el crecimiento se debe principalmente a mayor volumen y no necesariamente a mejoras en rentabilidad relativa.

### Q (Pregunta)
¿Qué factores del negocio como **categorías de producto, países o patrones temporales** están impulsando este crecimiento en ingresos y volumen de ventas?

### A (Respuesta)
El análisis general del dashboard muestra que:

- **Perú y Chile concentran la mayor parte de los ingresos**, siendo los mercados más relevantes para el negocio.
- Las **cuatro categorías de producto presentan un aporte relativamente equilibrado**, lo que indica una cartera diversificada.
- El comportamiento mensual muestra **patrones estacionales**, con caídas en mitad de año y recuperación hacia el último trimestre.

En conjunto, el crecimiento del negocio parece estar impulsado por **un aumento generalizado en la actividad comercial más que por cambios drásticos en precios o rentabilidad**.

## 🔎 Vista Detalle

### S (Situación)
Para entender con mayor profundidad los factores que explican el crecimiento, se analizan variables como **estacionalidad, segmento de cliente, país y nivel de venta**.

### C (Complicación)
El análisis detallado muestra diferencias importantes en el comportamiento del negocio:

- **La temporada de verano concentra el mayor nivel de ingresos**, mientras que invierno presenta menor actividad comercial.
- Los **clientes Premium y Estándar generan la mayor proporción del margen bruto** del negocio.
- Existen diferencias en el **nivel de ventas entre países**, lo que refleja distintos niveles de demanda regional.
- El **porcentaje de margen se mantiene relativamente consistente entre categorías**, lo que indica una estrategia de precios relativamente uniforme.

### Q (Pregunta)
¿Cómo puede la empresa aprovechar estos patrones para **seguir impulsando el crecimiento del negocio y maximizar la rentabilidad**?

### A (Respuesta)
A partir del análisis se identifican varias oportunidades estratégicas:

- **Potenciar campañas comerciales en temporadas de menor demanda**, especialmente durante invierno.
- **Priorizar estrategias de fidelización para clientes Premium**, ya que generan mayor valor para el negocio.
- Continuar fortaleciendo los mercados con mayor participación, como **Perú y Chile**.
- Mantener la **diversificación de categorías**, ya que todas contribuyen de manera equilibrada al ingreso total.

Estas acciones pueden ayudar a **sostener el crecimiento comercial y mejorar la eficiencia del negocio en el futuro**.
