````python
import pandas as pd
import numpy as np
import os

# Crear directorio
os.makedirs('data/dimensions', exist_ok=True)

# =========================================
# DIM PRODUCT
# =========================================

dim_product = pd.DataFrame([

    ["CO-600","Cola","600 ml","Soft Drink","Bottle",60,1400,3500,5000,180,0.015,"COP"],
    ["CO-1L","Cola","1 L","Soft Drink","Bottle",80,2200,5500,4000,180,0.015,"COP"],
    ["CO-2L","Cola","2 L","Soft Drink","Bottle",98,3600,9000,2500,180,0.015,"COP"],

    ["OR-600","Orange","600 ml","Soft Drink","Bottle",60,1400,3500,5000,180,0.015,"COP"],
    ["OR-1L","Orange","1 L","Soft Drink","Bottle",82,2200,5500,4000,180,0.015,"COP"],
    ["OR-2L","Orange","2 L","Soft Drink","Bottle",100,3600,9000,2500,180,0.015,"COP"],

    ["LE-600","Lemon Lime","600 ml","Soft Drink","Bottle",60,1400,3500,5000,180,0.015,"COP"],
    ["LE-1L","Lemon Lime","1 L","Soft Drink","Bottle",80,2200,5500,4000,180,0.015,"COP"],
    ["LE-2L","Lemon Lime","2 L","Soft Drink","Bottle",100,3600,9000,2500,180,0.015,"COP"],

    ["DC-600","Diet Cola","600 ml","Soft Drink","Bottle",65,1500,3800,4800,180,0.018,"COP"],
    ["DC-1L","Diet Cola","1 L","Soft Drink","Bottle",85,2400,6000,3800,180,0.018,"COP"],
    ["DC-2L","Diet Cola","2 L","Soft Drink","Bottle",105,3900,9800,2300,180,0.018,"COP"],

    ["EN-330","Energy Drink","330 ml","Energy Drink","Can",55,2800,6500,6000,365,0.020,"COP"],
    ["EN-500","Energy Drink","500 ml","Energy Drink","Can",70,3500,8500,5000,365,0.020,"COP"],

    ["TE-500","Tea","500 ml","Tea","Bottle",60,1700,4200,4500,120,0.018,"COP"],
    ["TE-1L","Tea","1 L","Tea","Bottle",82,2600,6500,3000,120,0.018,"COP"],

    ["WA-600","Water","600 ml","Water","Bottle",50,900,2200,5500,365,0.010,"COP"],
    ["WA-1L","Water","1 L","Water","Bottle",70,1300,3200,4000,365,0.010,"COP"],
    ["WA-2L","Water","2 L","Water","Bottle",90,2200,5000,2500,365,0.010,"COP"],
    ["WA-5L","Water","5 L","Water","Gallon",120,4500,9500,1500,365,0.010,"COP"]

],
columns=[
    "ProductID",
    "ProductFamily",
    "Size",
    "ProductCategory",
    "PackageType",
    "StandardBatchTimeMin",
    "UnitCost",
    "SellingPrice",
    "UnitsPerBatch",
    "ShelfLifeDays",
    "TargetScrapRate",
    "Currency"
])

# =========================================
# DIM OPERATOR
# =========================================

dim_operator = pd.DataFrame([

    ["OP001","Carlos Andres","Senior","2018-03-15",28,"Production","Level III",0.95,40],
    ["OP002","Juan David","Senior","2017-06-20",29,"Production","Level III",0.95,40],
    ["OP003","Luis Fernando","Senior","2019-01-10",30,"Production","Level III",0.95,40],
    ["OP004","Andres Felipe","Senior","2018-11-05",28,"Production","Level III",0.95,40],

    ["OP005","Jhon Alexander","Mid","2021-04-18",24,"Production","Level II",0.90,32],
    ["OP006","Sergio Andres","Mid","2021-09-12",25,"Production","Level II",0.90,32],
    ["OP007","Diego Alejandro","Mid","2022-02-25",24,"Production","Level II",0.90,32],
    ["OP008","Camilo Andres","Mid","2020-07-08",25,"Production","Level II",0.90,32],

    ["OP009","Julian Esteban","Junior","2023-01-15",20,"Production","Level I",0.85,24],
    ["OP010","Kevin Sebastian","Junior","2023-05-22",21,"Production","Level I",0.85,24],
    ["OP011","Mateo Santiago","Junior","2023-08-14",20,"Production","Level I",0.85,24],
    ["OP012","Brayan Steven","Junior","2024-02-01",19,"Production","Level I",0.85,24]

],
columns=[
    "OperatorID",
    "Operator",
    "ExperienceLevel",
    "HireDate",
    "HourlyRate",
    "Department",
    "CertificationLevel",
    "LaborEfficiencyTarget",
    "TrainingHoursYear"
])

# Convertir fecha de contratación a datetime
dim_operator["HireDate"] = pd.to_datetime(
    dim_operator["HireDate"]
)

# =========================================
# DIM LINE
# =========================================

dim_line = pd.DataFrame([

    ["L1",
     "Line A",
     6000,
     0.85,
     "2018-01-01",
     "High Speed PET",
     3,
     30],

    ["L2",
     "Line B",
     5000,
     0.82,
     "2020-06-15",
     "Standard PET",
     2,
     45],

    ["L3",
     "Line C",
     4000,
     0.80,
     "2022-09-01",
     "Flexible Packaging",
     1,
     60]

],
columns=[
    "LineID",
    "LineName",
    "MaxUnitsPerHour",
    "OEE_Target",
    "CommissionDate",
    "Technology",
    "CriticalityLevel",
    "MaintenanceFrequencyDays"
])

# Convertir fecha de puesta en marcha
dim_line["CommissionDate"] = pd.to_datetime(
    dim_line["CommissionDate"]
)

# =========================================
# DIM SHIFT
# =========================================

dim_shift = pd.DataFrame([

    ["A",
     "Morning Shift",
     "06:00",
     "14:00",
     8,
     "Andrea"],

    ["B",
     "Afternoon Shift",
     "14:00",
     "22:00",
     8,
     "German"],

    ["C",
     "Night Shift",
     "22:00",
     "06:00",
     8,
     "Hector"]

],
columns=[
    "ShiftID",
    "ShiftName",
    "ShiftStart",
    "ShiftEnd",
    "ShiftHours",
    "TeamLead"
])

# Convertir horas a formato datetime
dim_shift["ShiftStart"] = pd.to_datetime(
    dim_shift["ShiftStart"],
    format="%H:%M"
).dt.time

dim_shift["ShiftEnd"] = pd.to_datetime(
    dim_shift["ShiftEnd"],
    format="%H:%M"
).dt.time

# =========================================
# DIM DOWNTIME FACTOR
# =========================================

dim_downtime_factor = pd.DataFrame([

    [1,
     "Emergency stop",
     "No",
     "Safety",
     "High",
     15],

    [2,
     "Batch change",
     "Yes",
     "Operational",
     "Low",
     20],

    [3,
     "Labeling error",
     "No",
     "Quality",
     "Low",
     10],

    [4,
     "Inventory shortage",
     "No",
     "Supply Chain",
     "Medium",
     30],

    [5,
     "Product spill",
     "Yes",
     "Operational",
     "Medium",
     15],

    [6,
     "Machine adjustment",
     "Yes",
     "Operational",
     "Low",
     15],

    [7,
     "Machine failure",
     "No",
     "Mechanical",
     "High",
     60],

    [8,
     "Batch coding error",
     "Yes",
     "Quality",
     "Medium",
     15],

    [9,
     "Conveyor belt jam",
     "No",
     "Mechanical",
     "Medium",
     20],

    [10,
     "Calibration error",
     "Yes",
     "Quality",
     "Medium",
     25],

    [11,
     "Label switch",
     "Yes",
     "Operational",
     "Low",
     10],

    [12,
     "Other",
     "No",
     "Other",
     "Low",
     15]

],
columns=[
    "FactorID",
    "Description",
    "OperatorError",
    "Category",
    "Severity",
    "MTTR_Target"
])
# =========================================
# DIM DATE
# =========================================

import holidays

# Festivos oficiales de Colombia
co_holidays = holidays.Colombia(years=[2024, 2025])

# Calendario
dates = pd.date_range(
    start="2024-01-01",
    end="2025-06-30",
    freq="D"
)

dim_date = pd.DataFrame({
    "Date": dates
})

# DateKey
dim_date["DateKey"] = (
    dim_date["Date"]
    .dt.strftime("%Y%m%d")
    .astype(int)
)

# Componentes de fecha
dim_date["Year"] = dim_date["Date"].dt.year
dim_date["Quarter"] = dim_date["Date"].dt.quarter
dim_date["Month"] = dim_date["Date"].dt.month
dim_date["MonthName"] = dim_date["Date"].dt.month_name()
dim_date["MonthYear"]=dim_date["Date"].dt.strftime("%Y-%m")
dim_date["MonthYearName"]=dim_date["Date"].dt.strftime("%b-%Y")
dim_date["Semester"]=np.where(
    dim_date["Month"]<=6,
    1,
    2
)
dim_date["WeekDayNumber"] = (
    dim_date["Date"].dt.weekday + 1
)
# Semana ISO
dim_date["Week"] = dim_date["Date"].dt.isocalendar().week

# Día
dim_date["Day"] = dim_date["Date"].dt.day
dim_date["DayName"] = dim_date["Date"].dt.day_name()

# Día del año
dim_date["DayOfYear"] = dim_date["Date"].dt.dayofyear

# Fin de semana
dim_date["WeekendFlag"] = (
    dim_date["DayName"]
    .isin(["Sunday"])
)

# Festivos de Colombia
holiday_dates = pd.to_datetime(list(co_holidays.keys()))

dim_date["IsHoliday"] = dim_date["Date"].isin(
    holiday_dates
)

# Nombre del festivo
holiday_dict = {
    pd.Timestamp(date): name
    for date, name in co_holidays.items()
}

dim_date["HolidayName"] = dim_date["Date"].map(
    holiday_dict
)
# Identificar dias laborales
dim_date["WorkingDayFlag"] = ~(
    dim_date["WeekendFlag"]
    | dim_date["IsHoliday"]
)
# Fin de mes
dim_date["IsMonthEnd"] = dim_date["Date"].dt.is_month_end

# Fin de trimestre
dim_date["IsQuarterEnd"] = dim_date["Date"].dt.is_quarter_end

# Reordenar columnas
dim_date = dim_date[
    [
        "DateKey",
        "Date",
        "Year",
        "Semester",
        "Quarter",
        "MonthYear",
        "MonthYearName",
        "Month",
        "MonthName",
        "Week",
        "WeekDayNumber",
        "Day",
        "DayName",
        "DayOfYear",
        "WeekendFlag",
        "IsHoliday",
        "HolidayName",
        "IsMonthEnd",
        "IsQuarterEnd",
        "WorkingDayFlag"
    ]
]

# =========================================
# EXPORTAR CSV
# =========================================

dim_product.to_csv(
    "data/dimensions/dim_product.csv",
    index=False
)

dim_operator.to_csv(
    "data/dimensions/dim_operator.csv",
    index=False
)

dim_line.to_csv(
    "data/dimensions/dim_line.csv",
    index=False
)

dim_shift.to_csv(
    "data/dimensions/dim_shift.csv",
    index=False
)

dim_downtime_factor.to_csv(
    "data/dimensions/dim_downtime_factor.csv",
    index=False
)

dim_date.to_csv(
    "data/dimensions/dim_date.csv",
    index=False
)

print("Dimensions created successfully.")
````

# ⚙️ Construcción de FactProduction

La tabla **FactProduction** es la tabla de hechos principal del modelo y representa cada lote producido en la planta.

Su construcción se realiza en varias etapas con el fin de simular el comportamiento real de una operación manufacturera.

---

## Generación de llaves y relaciones

Se crean los registros de producción y se asignan las dimensiones correspondientes:

- Fecha.
- Producto.
- Operador.
- Línea de producción.
- Turno.

Cada registro representa un lote de producción único.

---

## Simulación de tiempos de producción

Se calculan:

- Tiempo estándar del lote.
- Tiempo de parada (Downtime).
- Tiempo real de producción.

Los tiempos de parada dependen del nivel de experiencia del operador, simulando condiciones operativas reales.

---

## Generación de cantidades producidas

Se calculan:

- Unidades producidas.
- Scrap.
- Unidades reprocesadas.
- Unidades buenas.

La productividad y la tasa de desperdicio se ven influenciadas por la experiencia del operador.

---

## Cálculo de costos y rentabilidad

Se incorporan métricas económicas como:

- Costos de producción.
- Costos de mano de obra.
- Costos de scrap.
- Costos de reproceso.
- Costos de mantenimiento.
- Costos energéticos.
- Ingresos.
- Utilidad bruta.
- Margen de beneficio.

---

## Indicadores de calidad y productividad

Se generan indicadores operacionales como:

- Scrap Rate.
- Rework Rate.
- First Pass Yield.
- Labor Efficiency.
- Units per Hour.
- Units per Labor Hour.

---

## Indicadores OEE

Se calculan los componentes del Overall Equipment Effectiveness:

- Availability.
- Performance.
- Quality.
- OEE.

Adicionalmente se generan:

- Downtime Rate.
- Indicadores World Class.
- Clasificación del OEE.

---

## Resultado

La tabla FactProduction constituye el núcleo del modelo analítico y servirá como base para la generación de las demás tablas de hechos relacionadas con:

- Calidad.
- Downtime.
- Inventarios.
- Logística.
- Seguridad industrial.
- Mantenimiento.

Finalmente, los datos son exportados a formato CSV para su posterior integración en Power BI.

````python
# ==========================================
# FACT PRODUCTION
# STEP 1
# GENERATE KEYS
# ==========================================

import pandas as pd
import numpy as np
import os

np.random.seed(42)

# Crear directorio
os.makedirs("data/facts", exist_ok=True)

# ==========================================
# LOAD DIMENSIONS
# ==========================================

dim_date = pd.read_csv("data/dimensions/dim_date.csv")
dim_product = pd.read_csv("data/dimensions/dim_product.csv")
dim_operator = pd.read_csv("data/dimensions/dim_operator.csv")
dim_line = pd.read_csv("data/dimensions/dim_line.csv")

# Convertir fecha
dim_date["Date"] = pd.to_datetime(dim_date["Date"])

# ==========================================
# NUMBER OF BATCHES
# ==========================================

n_batches = 10000

# ==========================================
# BatchID
# ==========================================

batch_ids = [
    f"B{str(i).zfill(6)}"
    for i in range(1, n_batches + 1)
]

# ==========================================
# DateKey
# ==========================================

date_keys = np.random.choice(
    dim_date["DateKey"],
    size=n_batches
)

# ==========================================
# ProductID
# ==========================================

product_ids = np.random.choice(
    dim_product["ProductID"],
    size=n_batches
)

# ==========================================
# ShiftID
# ==========================================

shift_ids = np.random.choice(
    ["A", "B", "C"],
    size=n_batches,
    p=[0.40, 0.35, 0.25]
)

# ==========================================
# LineID
# ==========================================

line_ids = np.random.choice(
    ["L1", "L2", "L3"],
    size=n_batches,
    p=[0.45, 0.35, 0.20]
)

# ==========================================
# OPERATOR POOLS
# ==========================================

senior_ops = (
    dim_operator.loc[
        dim_operator["ExperienceLevel"] == "Senior",
        "OperatorID"
    ]
    .tolist()
)

mid_ops = (
    dim_operator.loc[
        dim_operator["ExperienceLevel"] == "Mid",
        "OperatorID"
    ]
    .tolist()
)

junior_ops = (
    dim_operator.loc[
        dim_operator["ExperienceLevel"] == "Junior",
        "OperatorID"
    ]
    .tolist()
)

# ==========================================
# OperatorID according to Line
# ==========================================

operator_ids = []

for line in line_ids:

    if line == "L1":

        operator_pool = (
            senior_ops +
            mid_ops
        )

    elif line == "L2":

        operator_pool = (
            senior_ops +
            mid_ops +
            junior_ops
        )

    else:

        operator_pool = (
            mid_ops +
            junior_ops
        )

    operator_ids.append(
        np.random.choice(operator_pool)
    )

# ==========================================
# CREATE FACT TABLE
# ==========================================

fact_production = pd.DataFrame({

    "BatchID": batch_ids,
    "DateKey": date_keys,
    "ProductID": product_ids,
    "OperatorID": operator_ids,
    "LineID": line_ids,
    "ShiftID": shift_ids

})

# ==========================================
# SORT
# ==========================================

fact_production = (

    fact_production
    .sort_values(
        ["DateKey", "LineID"]
    )
    .reset_index(drop=True)

)

# ==========================================
# STEP 2
# TIME VARIABLES
# ==========================================

# ==========================================
# Merge Standard Batch Time
# ==========================================

fact_production = fact_production.merge(

    dim_product[
        [
            "ProductID",
            "StandardBatchTimeMin"
        ]
    ],

    on="ProductID",
    how="left"

)

# ==========================================
# Planned Batch Time
# ==========================================

fact_production["PlannedBatchTimeMin"] = (

    fact_production["StandardBatchTimeMin"]

)

# ==========================================
# Merge Experience Level
# ==========================================

fact_production = fact_production.merge(

    dim_operator[
        [
            "OperatorID",
            "ExperienceLevel"
        ]
    ],

    on="OperatorID",
    how="left"

)

# ==========================================
# Downtime by Experience Level
# ==========================================

np.random.seed(42)

downtime = []

for level in fact_production["ExperienceLevel"]:

    if level == "Senior":

        dt = np.random.exponential(
            scale=8
        )

    elif level == "Mid":

        dt = np.random.exponential(
            scale=14
        )

    else:

        dt = np.random.exponential(
            scale=20
        )

    downtime.append(dt)

# ==========================================
# DowntimeMin
# ==========================================

fact_production["DowntimeMin"] = (

    np.array(downtime)
    .clip(
        0,
        120
    )
    .round()

)

# ==========================================
# Actual Batch Time
# ==========================================

fact_production["ActualBatchTimeMin"] = (

    fact_production["PlannedBatchTimeMin"]

    +

    fact_production["DowntimeMin"]

)

# ==========================================
# Remove Auxiliary Columns
# ==========================================

fact_production.drop(

    columns=[
        "StandardBatchTimeMin",
        "ExperienceLevel"
    ],

    inplace=True

)

# ==========================================
# CHECK
# ==========================================

print(fact_production.head())

print()
print("Rows:", len(fact_production))

# ==========================================
# STEP 3
# GENERATE PRODUCTION QUANTITIES
# ==========================================

# Recuperar información del producto
fact_production = fact_production.merge(

    dim_product[
        [
            "ProductID",
            "UnitsPerBatch",
            "TargetScrapRate"
        ]
    ],

    on="ProductID",
    how="left"

)

# Recuperar ExperienceLevel nuevamente
fact_production = fact_production.merge(

    dim_operator[
        [
            "OperatorID",
            "ExperienceLevel"
        ]
    ],

    on="OperatorID",
    how="left"

)

# ==========================================
# Units Produced
# efecto del nivel de experiencia
# ==========================================

np.random.seed(42)

units_produced = []

for _, row in fact_production.iterrows():

    base_units = row["UnitsPerBatch"]

    if row["ExperienceLevel"] == "Senior":

        factor = np.random.normal(
            loc=1.02,
            scale=0.02
        )

    elif row["ExperienceLevel"] == "Mid":

        factor = np.random.normal(
            loc=1.00,
            scale=0.03
        )

    else:

        factor = np.random.normal(
            loc=0.97,
            scale=0.04
        )

    units = int(base_units * factor)

    units_produced.append(
        max(units, 1)
    )

fact_production["UnitsProduced"] = units_produced

# ==========================================
# Scrap Units
# ==========================================

scrap_units = []

for _, row in fact_production.iterrows():

    scrap_rate = row["TargetScrapRate"]

    # efecto del operador
    if row["ExperienceLevel"] == "Senior":

        scrap_rate *= 0.80

    elif row["ExperienceLevel"] == "Mid":

        scrap_rate *= 1.00

    else:

        scrap_rate *= 1.30

    scrap = np.random.poisson(

        row["UnitsProduced"]
        * scrap_rate

    )

    scrap_units.append(scrap)

fact_production["ScrapUnits"] = scrap_units

# ==========================================
# Rework Units
# 30% del scrap es recuperable
# ==========================================

fact_production["ReworkUnits"] = np.random.poisson(

    fact_production["ScrapUnits"] * 0.30

)

# ==========================================
# Good Units
# ==========================================

fact_production["GoodUnits"] = (

      fact_production["UnitsProduced"]
    - fact_production["ScrapUnits"]
    + fact_production["ReworkUnits"]

)

# Evitar negativos
fact_production["GoodUnits"] = (

    fact_production["GoodUnits"]
    .clip(lower=0)

)

# ==========================================
# Eliminar columnas auxiliares
# ==========================================

fact_production.drop(

    columns=[
        "UnitsPerBatch",
        "TargetScrapRate",
        "ExperienceLevel"
    ],

    inplace=True

)

# ==========================================
# Verificación
# ==========================================

print(

    fact_production[
        [
            "ProductID",
            "UnitsProduced",
            "ScrapUnits",
            "ReworkUnits",
            "GoodUnits"
        ]
    ].head()

)

# ==========================================
# STEP 4
# COSTS & REVENUE
# ==========================================

# Recuperar información del producto
fact_production = fact_production.merge(

    dim_product[
        [
            "ProductID",
            "UnitCost",
            "SellingPrice"
        ]
    ],

    on="ProductID",
    how="left"

)

# Recuperar tarifa horaria del operador
fact_production = fact_production.merge(

    dim_operator[
        [
            "OperatorID",
            "HourlyRate"
        ]
    ],

    on="OperatorID",
    how="left"

)

# ==========================================
# Labor Hours
# ==========================================

fact_production["LaborHours"] = (

    fact_production["ActualBatchTimeMin"]
    / 60

)

# ==========================================
# Labor Cost
# ==========================================

fact_production["LaborCost"] = (

    fact_production["LaborHours"]
    *
    fact_production["HourlyRate"]

)

# ==========================================
# Production Cost
# ==========================================

fact_production["ProductionCost"] = (

    fact_production["UnitsProduced"]
    *
    fact_production["UnitCost"]

)

# ==========================================
# Scrap Cost
# ==========================================

fact_production["ScrapCost"] = (

    fact_production["ScrapUnits"]
    *
    fact_production["UnitCost"]

)

# ==========================================
# Rework Cost
# ==========================================

fact_production["ReworkCost"] = (

    fact_production["ReworkUnits"]
    *
    fact_production["UnitCost"]
    * 0.40

)

# ==========================================
# Maintenance Cost
# ==========================================

fact_production["MaintenanceCost"] = (

    fact_production["DowntimeMin"]
    * 3000

)

# ==========================================
# Energy Consumption
# ==========================================

fact_production["EnergyConsumptionKWh"] = (

    fact_production["ActualBatchTimeMin"]

    *

    np.random.uniform(
        2.5,
        4.0,
        len(fact_production)
    )

)

# ==========================================
# Energy Cost
# ==========================================

fact_production["EnergyCost"] = (

    fact_production["EnergyConsumptionKWh"]
    * 750

)

# ==========================================
# Water Consumption
# ==========================================

fact_production["WaterConsumptionLiters"] = (

    fact_production["UnitsProduced"]

    *

    np.random.uniform(
        0.8,
        1.3,
        len(fact_production)
    )

)

# ==========================================
# CO2 Emissions
# ==========================================

fact_production["CO2EmissionKg"] = (

    fact_production["EnergyConsumptionKWh"]
    * 0.18

)

# ==========================================
# Revenue
# ==========================================

fact_production["Revenue"] = (

    fact_production["GoodUnits"]
    *
    fact_production["SellingPrice"]

)

# ==========================================
# Total Manufacturing Cost
# ==========================================

fact_production["TotalManufacturingCost"] = (

      fact_production["ProductionCost"]
    + fact_production["LaborCost"]
    + fact_production["ScrapCost"]
    + fact_production["ReworkCost"]
    + fact_production["MaintenanceCost"]
    + fact_production["EnergyCost"]

)

# ==========================================
# Gross Profit
# ==========================================

fact_production["GrossProfit"] = (

    fact_production["Revenue"]

    -

    fact_production["TotalManufacturingCost"]

)

# ==========================================
# Profit Margin %
# ==========================================

fact_production["ProfitMarginPct"] = np.where(

    fact_production["Revenue"] > 0,

    fact_production["GrossProfit"]
    /
    fact_production["Revenue"],

    0

)

# ==========================================
# Cost Per Unit
# ==========================================

fact_production["CostPerUnit"] = np.where(

    fact_production["GoodUnits"] > 0,

    fact_production["TotalManufacturingCost"]
    /
    fact_production["GoodUnits"],

    0

)

# ==========================================
# Eliminar columnas auxiliares
# ==========================================

fact_production.drop(

    columns=[
        "UnitCost",
        "SellingPrice",
        "HourlyRate"
    ],

    inplace=True

)

# ==========================================
# Redondear métricas
# ==========================================

cols_round = [

    "LaborHours",
    "LaborCost",
    "ProductionCost",
    "ScrapCost",
    "ReworkCost",
    "MaintenanceCost",
    "EnergyConsumptionKWh",
    "EnergyCost",
    "WaterConsumptionLiters",
    "CO2EmissionKg",
    "Revenue",
    "TotalManufacturingCost",
    "GrossProfit",
    "ProfitMarginPct",
    "CostPerUnit"

]

fact_production[cols_round] = (

    fact_production[cols_round]
    .round(2)

)

# ==========================================
# Verificación
# ==========================================

print(

    fact_production[
        [
            "Revenue",
            "TotalManufacturingCost",
            "GrossProfit",
            "ProfitMarginPct",
            "CostPerUnit"
        ]
    ].head()

)

# ==========================================
# STEP 5
# QUALITY & LABOR KPIs
# ==========================================

# ------------------------------------------
# Scrap Rate
# % de unidades rechazadas
# ------------------------------------------

fact_production["ScrapRate"] = np.where(

    fact_production["UnitsProduced"] > 0,

    fact_production["ScrapUnits"]
    /
    fact_production["UnitsProduced"],

    0

)

# ------------------------------------------
# Rework Rate
# % de unidades retrabajadas
# ------------------------------------------

fact_production["ReworkRate"] = np.where(

    fact_production["UnitsProduced"] > 0,

    fact_production["ReworkUnits"]
    /
    fact_production["UnitsProduced"],

    0

)

# ------------------------------------------
# First Pass Yield (FPY)
# % de unidades buenas sin retrabajo
# ------------------------------------------

fact_production["FirstPassYield"] = np.where(

    fact_production["UnitsProduced"] > 0,

    (

        fact_production["GoodUnits"]
        -
        fact_production["ReworkUnits"]

    )

    /

    fact_production["UnitsProduced"],

    0

)

# ------------------------------------------
# Labor Efficiency
# compara tiempo estándar vs tiempo real
# ------------------------------------------

fact_production["LaborEfficiency"] = np.where(

    fact_production["ActualBatchTimeMin"] > 0,

    fact_production["PlannedBatchTimeMin"]

    /

    fact_production["ActualBatchTimeMin"],

    0

)

# ------------------------------------------
# Units per Hour
# productividad real de la línea
# ------------------------------------------

fact_production["UnitsPerHour"] = np.where(

    fact_production["ActualBatchTimeMin"] > 0,

    fact_production["UnitsProduced"]

    /

    (

        fact_production["ActualBatchTimeMin"]
        / 60

    ),

    0

)

# ------------------------------------------
# Units per Labor Hour
# productividad de mano de obra
# ------------------------------------------

fact_production["UnitsPerLaborHour"] = np.where(

    fact_production["LaborHours"] > 0,

    fact_production["GoodUnits"]

    /

    fact_production["LaborHours"],

    0

)

# ------------------------------------------
# Yield %
# porcentaje de producto bueno
# ------------------------------------------

fact_production["YieldPct"] = np.where(

    fact_production["UnitsProduced"] > 0,

    fact_production["GoodUnits"]

    /

    fact_production["UnitsProduced"],

    0

)

# ------------------------------------------
# Loss Units
# unidades perdidas definitivamente
# ------------------------------------------

fact_production["LossUnits"] = (

    fact_production["UnitsProduced"]

    -

    fact_production["GoodUnits"]

)

# ------------------------------------------
# Loss Cost
# costo asociado a las pérdidas
# ------------------------------------------

fact_production["LossCost"] = (

    fact_production["LossUnits"]

    *

    fact_production["CostPerUnit"]

)

# ------------------------------------------
# Redondear métricas
# ------------------------------------------

cols_round = [

    "ScrapRate",
    "ReworkRate",
    "FirstPassYield",
    "LaborEfficiency",
    "UnitsPerHour",
    "UnitsPerLaborHour",
    "YieldPct",
    "LossCost"

]

fact_production[cols_round] = (

    fact_production[cols_round]
    .round(4)

)

# ==========================================
# Verificación
# ==========================================

print(

    fact_production[
        [
            "ScrapRate",
            "ReworkRate",
            "FirstPassYield",
            "LaborEfficiency",
            "UnitsPerHour",
            "UnitsPerLaborHour",
            "YieldPct"
        ]
    ].head()

)

# ==========================================
# STEP 6
# OEE & RELIABILITY KPIs
# ==========================================

# ==========================================
# Recuperar capacidad de la línea
# ==========================================

fact_production = fact_production.merge(

    dim_line[
        [
            "LineID",
            "MaxUnitsPerHour"
        ]
    ],

    on="LineID",
    how="left"

)

# ==========================================
# Planned Production Time
# ==========================================

fact_production["PlannedProductionTimeMin"] = (

    fact_production["PlannedBatchTimeMin"]

)

# ==========================================
# Operating Time
# ==========================================

fact_production["OperatingTimeMin"] = (

    fact_production["PlannedProductionTimeMin"]

    -

    fact_production["DowntimeMin"]

)

# Evitar valores negativos

fact_production["OperatingTimeMin"] = (

    fact_production["OperatingTimeMin"]
    .clip(lower=0)

)

# ==========================================
# Availability
# ==========================================

fact_production["Availability"] = np.where(

    fact_production["PlannedProductionTimeMin"] > 0,

    fact_production["OperatingTimeMin"]

    /

    fact_production["PlannedProductionTimeMin"],

    0

)

# ==========================================
# Ideal Units
# producción teórica durante el tiempo real
# ==========================================

fact_production["IdealUnits"] = (

    fact_production["MaxUnitsPerHour"]

    *

    (

        fact_production["OperatingTimeMin"]

        / 60

    )

)

# ==========================================
# Performance
# ==========================================

fact_production["Performance"] = np.where(

    fact_production["IdealUnits"] > 0,

    fact_production["UnitsProduced"]

    /

    fact_production["IdealUnits"],

    0

)

# Limitar máximo 105%

fact_production["Performance"] = (

    fact_production["Performance"]
    .clip(upper=1.05)

)

# ==========================================
# Quality
# ==========================================

fact_production["Quality"] = np.where(

    fact_production["UnitsProduced"] > 0,

    fact_production["GoodUnits"]

    /

    fact_production["UnitsProduced"],

    0

)

# ==========================================
# OEE
# ==========================================

fact_production["OEE"] = (

    fact_production["Availability"]

    *

    fact_production["Performance"]

    *

    fact_production["Quality"]

)

# ==========================================
# Downtime Rate
# ==========================================

fact_production["DowntimeRate"] = np.where(

    fact_production["PlannedProductionTimeMin"] > 0,

    fact_production["DowntimeMin"]

    /

    fact_production["PlannedProductionTimeMin"],

    0

)

# ==========================================
# Breakdown Flag
# ==========================================

fact_production["BreakdownFlag"] = np.where(

    fact_production["DowntimeMin"] > 0,

    1,

    0

)

# ==========================================
# World Class Flags
# ==========================================

fact_production["WorldClassAvailability"] = np.where(

    fact_production["Availability"] >= 0.90,

    1,

    0

)

fact_production["WorldClassPerformance"] = np.where(

    fact_production["Performance"] >= 0.95,

    1,

    0

)

fact_production["WorldClassQuality"] = np.where(

    fact_production["Quality"] >= 0.99,

    1,

    0

)

fact_production["WorldClassOEE"] = np.where(

    fact_production["OEE"] >= 0.85,

    1,

    0

)

# ==========================================
# OEE Category
# ==========================================

fact_production["OEE_Category"] = np.select(

    [

        fact_production["OEE"] >= 0.85,

        fact_production["OEE"] >= 0.70,

        fact_production["OEE"] >= 0.50

    ],

    [

        "World Class",

        "Good",

        "Average"

    ],

    default="Poor"

)

# ==========================================
# Redondear KPIs
# ==========================================

cols_round = [

    "Availability",
    "Performance",
    "Quality",
    "OEE",
    "DowntimeRate"

]

fact_production[cols_round] = (

    fact_production[cols_round]
    .round(4)

)

# ==========================================
# Eliminar columnas auxiliares
# ==========================================

fact_production.drop(

    columns=[

        "MaxUnitsPerHour"

    ],

    inplace=True

)

# ==========================================
# Verificación
# ==========================================

print(

    fact_production[
        [
            "Availability",
            "Performance",
            "Quality",
            "OEE",
            "DowntimeRate",
            "OEE_Category"
        ]
    ].head()

)

print()
print("Rows:", len(fact_production))

# ==========================================
# EXPORTAR CSV
# ==========================================

fact_production.to_csv(

    "data/facts/fact_production.csv",

    index=False

)

print()
print("fact_production.csv created successfully.")
````

````python
# ==========================================
# FACT INVENTORY
# ==========================================

import numpy as np
import pandas as pd

# reproducibilidad
np.random.seed(42)

inventory_rows = []

inventory_counter = 1

# inventario inicial por producto
current_inventory = {}

for _, product in dim_product.iterrows():

    current_inventory[product["ProductID"]] = np.random.randint(
        10000,
        30000
    )

# ==================================================
# LOOP SOBRE FACT PRODUCTION
# ==================================================

for _, row in fact_production.iterrows():

    product_info = dim_product[
        dim_product["ProductID"] == row["ProductID"]
    ].iloc[0]

    product_id = row["ProductID"]

    beginning_inventory = current_inventory[product_id]

    # ======================================
    # DEMANDA
    # ======================================

    demand_units = int(

        row["GoodUnits"]

        *

        np.random.uniform(
            0.90,
            1.15
        )

    )

    # ======================================
    # SHIPPED UNITS
    # ======================================

    shipped_units = min(

        beginning_inventory
        +
        row["GoodUnits"],

        demand_units

    )

    # ======================================
    # ENDING INVENTORY
    # ======================================

    ending_inventory = (

        beginning_inventory

        +

        row["GoodUnits"]

        -

        shipped_units

    )

    # actualizar inventario para siguiente batch
    current_inventory[product_id] = ending_inventory

    # ======================================
    # SAFETY STOCK
    # ======================================

    safety_stock = int(

        product_info["UnitsPerBatch"]

        * 1.5

    )

    # ======================================
    # REORDER POINT
    # ======================================

    reorder_point = int(

        safety_stock

        +

        product_info["UnitsPerBatch"]

        * 2

    )

    # ======================================
    # BACKORDER
    # ======================================

    backorder_units = max(

        demand_units
        -
        shipped_units,

        0

    )

    stockout_flag = (

        backorder_units > 0

    )

    # ======================================
    # EXPIRATIONS
    # ======================================

    expired_units = np.random.binomial(

        ending_inventory,

        0.001

    )

    obsolete_units = np.random.binomial(

        ending_inventory,

        0.0005

    )

    # ======================================
    # INVENTORY VALUE
    # ======================================

    inventory_value = (

        ending_inventory

        *

        product_info["UnitCost"]

    )

    # ======================================
    # HOLDING COST
    # ======================================

    annual_holding_rate = 0.18

    holding_cost = (

        inventory_value

        *

        annual_holding_rate

        / 365

    )

    # ======================================
    # OBSOLETE COST
    # ======================================

    obsolete_cost = (

        obsolete_units

        *

        product_info["UnitCost"]

    )

    # ======================================
    # FILL RATE
    # ======================================

    fill_rate = (

        shipped_units

        /

        demand_units

    )

    # ======================================
    # INVENTORY TURNOVER
    # ======================================

    average_inventory = (

        beginning_inventory
        +
        ending_inventory

    ) / 2

    cogs = (

        shipped_units

        *

        product_info["UnitCost"]

    )

    inventory_turnover = (

        cogs

        /

        average_inventory

    )

    # ======================================
    # DAYS INVENTORY OUTSTANDING
    # ======================================

    dio = (

        365

        /

        inventory_turnover

    )

    # ======================================
    # APPEND
    # ======================================

    inventory_rows.append({

        # Keys
        "InventoryID":
        f"INV{inventory_counter:08d}",

        "DateKey":
        row["DateKey"],

        "ProductID":
        row["ProductID"],

        # Stock
        "BeginningInventory":
        round(beginning_inventory),

        "ProducedUnits":
        round(row["GoodUnits"]),

        "DemandUnits":
        round(demand_units),

        "ShippedUnits":
        round(shipped_units),

        "EndingInventory":
        round(ending_inventory),

        # Planning
        "SafetyStock":
        safety_stock,

        "ReorderPoint":
        reorder_point,

        # Shortages
        "BackorderUnits":
        round(backorder_units),

        "StockoutFlag":
        stockout_flag,

        # Obsolescence
        "ExpiredUnits":
        expired_units,

        "ObsoleteUnits":
        obsolete_units,

        # Costs
        "InventoryValue":
        round(inventory_value,2),

        "HoldingCost":
        round(holding_cost,2),

        "ObsoleteCost":
        round(obsolete_cost,2),

        # KPIs
        "InventoryTurnover":
        round(inventory_turnover,2),

        "DaysInventoryOutstanding":
        round(dio,2),

        "FillRate":
        round(fill_rate,4)

    })

    inventory_counter += 1

# ==================================================
# CREATE DATAFRAME
# ==================================================

fact_inventory = pd.DataFrame(
    inventory_rows
)

print(
    "Rows:",
    len(fact_inventory)
)

fact_inventory.head()
# =========================================
# EXPORTAR CSV
# =========================================

fact_inventory.to_csv(
    "data/facts/fact_inventory.csv",
    index=False
)
````

````python
# ==========================================
# FACT QUALITY
# ==========================================

import numpy as np

quality_rows = []

quality_counter = 1

# Distribución de defectos
defect_probs = np.array(
    [0.30, 0.25, 0.20, 0.15, 0.10]
)

# Severidad posible
severity_options = [
    "Low",
    "Medium",
    "High"
]

severity_probs = [
    0.55,
    0.35,
    0.10
]

# Tipo de defecto
origin_options = [
    "Internal",
    "External"
]

origin_probs = [
    0.92,
    0.08
]

# ==========================================
# LOOP
# ==========================================

for _, row in fact_production.iterrows():

    scrap_units = int(row["ScrapUnits"])

    rework_units = int(row["ReworkUnits"])

    total_defects = scrap_units + rework_units

    # =====================================
    # Distribución de defectos
    # =====================================

    if total_defects > 0:

        defect_split = np.random.multinomial(
            total_defects,
            defect_probs
        )

    else:

        defect_split = [0,0,0,0,0]

    label_defects = defect_split[0]

    packaging_defects = defect_split[1]

    filling_defects = defect_split[2]

    cap_defects = defect_split[3]

    appearance_defects = defect_split[4]

    # =====================================
    # Coste unitario aproximado
    # =====================================

    if scrap_units > 0:

        cost_per_unit = (
            row["ScrapCost"]
            /
            scrap_units
        )

    else:

        cost_per_unit = (
            row["ProductionCost"]
            /
            row["UnitsProduced"]
        )

    # =====================================
    # Rework Cost
    # =====================================

    rework_cost = (

        rework_units

        *

        cost_per_unit

        *

        np.random.uniform(
            0.25,
            0.50
        )

    )

    # =====================================
    # Inspection Cost
    # =====================================

    inspection_cost = (

        row["UnitsProduced"]

        *

        np.random.uniform(
            15,
            35
        )

    )

    # =====================================
    # Cost of Poor Quality
    # =====================================

    copq = (

        row["ScrapCost"]

        +

        rework_cost

        +

        inspection_cost

    )

    # =====================================
    # Yield Loss
    # =====================================

    yield_loss = (

        scrap_units

        /

        row["UnitsProduced"]

    )

    # =====================================
    # Defect Density
    # =====================================

    defect_density = (

        total_defects

        /

        row["UnitsProduced"]

    )

    # =====================================
    # DPMO
    # =====================================

    opportunities_per_unit = 5

    dpmo = (

        total_defects

        /

        (

            row["UnitsProduced"]

            *

            opportunities_per_unit

        )

    ) * 1_000_000

    # =====================================
    # Sigma Level
    # =====================================

    if dpmo <= 3.4:

        sigma_level = 6

    elif dpmo <= 233:

        sigma_level = 5

    elif dpmo <= 6210:

        sigma_level = 4

    elif dpmo <= 66807:

        sigma_level = 3

    elif dpmo <= 308537:

        sigma_level = 2

    else:

        sigma_level = 1

    # =====================================
    # RTY
    # =====================================

    n_process_steps = 5

    rty = (
        row["FirstPassYield"]
        ** n_process_steps
    )

    # =====================================
    # Defect Severity
    # =====================================

    defect_severity = np.random.choice(
        severity_options,
        p=severity_probs
    )

    # =====================================
    # Defect Origin
    # =====================================

    defect_origin = np.random.choice(
        origin_options,
        p=origin_probs
    )

    # =====================================
    # Categoría dominante
    # =====================================

    categories = {

        "Label":
        label_defects,

        "Packaging":
        packaging_defects,

        "Filling":
        filling_defects,

        "Cap":
        cap_defects,

        "Appearance":
        appearance_defects

    }

    defect_category = max(
        categories,
        key=categories.get
    )

    # =====================================
    # APPEND
    # =====================================

    quality_rows.append({

        "QualityID":
        f"QLT{quality_counter:08d}",

        "BatchID":
        row["BatchID"],

        "DateKey":
        row["DateKey"],

        "ProductID":
        row["ProductID"],

        "LineID":
        row["LineID"],

        "OperatorID":
        row["OperatorID"],

        "ShiftID":
        row["ShiftID"],

# Producción
"UnitsProduced": row["UnitsProduced"],
"GoodUnits": row["GoodUnits"],

# Defectos
"ScrapUnits": row["ScrapUnits"],
"ReworkUnits": row["ReworkUnits"],
"TotalDefects": row["ScrapUnits"] + row["ReworkUnits"],

        # Tipos de defectos
        "LabelDefects":
        label_defects,

        "PackagingDefects":
        packaging_defects,

        "FillingDefects":
        filling_defects,

        "CapDefects":
        cap_defects,

        "AppearanceDefects":
        appearance_defects,

        # Clasificación
        "DefectCategory":
        defect_category,

        "DefectSeverity":
        defect_severity,

        "DefectOrigin":
        defect_origin,

        # Costos
        "ScrapCost":
        round(row["ScrapCost"],2),

        "ReworkCost":
        round(rework_cost,2),

        "InspectionCost":
        round(inspection_cost,2),

        "CostOfPoorQuality":
        round(copq,2),

        # KPIs
        "ScrapRate":
        round(row["ScrapRate"],4),

        "ReworkRate":
        round(row["ReworkRate"],4),

        "FirstPassYield":
        round(row["FirstPassYield"],4),

        "YieldLoss":
        round(yield_loss,4),

        "DefectDensity":
        round(defect_density,4),

        "RolledThroughputYield":
        round(rty,4),

        "DefectsPerMillion":
        round(dpmo,2),

        "SigmaLevel":
        sigma_level

    })

    quality_counter += 1

# ==========================================
# CREATE DATAFRAME
# ==========================================

fact_quality = pd.DataFrame(
    quality_rows
)

print(
    "Rows:",
    len(fact_quality)
)

fact_quality.head()
# =========================================
# EXPORTAR CSV
# =========================================

fact_quality.to_csv(
    "data/facts/fact_quality.csv",
    index=False
)
````

# 🔧 Construcción de FactMaintenance

La tabla **FactMaintenance** almacena la información relacionada con las actividades de mantenimiento ejecutadas sobre las líneas de producción.

Cada registro representa una orden de mantenimiento y permite analizar la confiabilidad y mantenibilidad de los activos industriales.

---

## Generación de eventos de mantenimiento

Se simula la ocurrencia de mantenimientos a lo largo de la operación, incorporando diferentes estrategias de intervención:

- Preventivo.
- Correctivo.
- Predictivo.

También se identifica si el mantenimiento fue planificado o consecuencia de una falla.

---

## Tiempos de mantenimiento

Para cada intervención se generan variables asociadas al proceso de reparación:

- Horas de mantenimiento.
- Horas de parada.
- Tiempo Medio Entre Fallas (MTBF).
- Tiempo Medio de Reparación (MTTR).

---

## Costos de mantenimiento

Se estiman los costos asociados a:

- Mano de obra.
- Repuestos.
- Costo total de mantenimiento.

Estas variables permiten evaluar el impacto económico de las intervenciones sobre la operación.

---

## Indicadores de confiabilidad

Se incorporan métricas utilizadas en Ingeniería de Confiabilidad y Asset Management:

- Failure Rate (λ).
- Operating Time.
- Reliability \(R(t)=e^{-t/MTBF}\).
- Maintainability \(M(t)=1-e^{-t/MTTR}\).

Estos indicadores permiten cuantificar la probabilidad de operación sin fallas y la capacidad de recuperación del sistema.

---

## Indicadores de desempeño del mantenimiento

La tabla también incorpora métricas relacionadas con la gestión del área:

- Maintenance Efficiency.
- Preventive Compliance.
- Failure Flag.

Estas variables permiten analizar la efectividad de las estrategias de mantenimiento y el nivel de cumplimiento de las actividades preventivas.

---

## Resultado

La tabla FactMaintenance permite analizar:

- Costos de mantenimiento.
- Distribución entre mantenimiento preventivo, correctivo y predictivo.
- Confiabilidad y mantenibilidad de los equipos.
- Frecuencia de fallas.
- Eficiencia de las intervenciones.
- Cumplimiento del plan de mantenimiento.

Finalmente, la información es exportada a formato CSV para su posterior análisis en Power BI.

````python
# ==========================================
# FACT MAINTENANCE
# ==========================================

import numpy as np

maintenance_rows = []

maintenance_counter = 1

maintenance_types = [
    "Preventive",
    "Corrective",
    "Predictive"
]

maintenance_probs = [
    0.55,
    0.35,
    0.10
]

# ==========================================
# LOOP
# ==========================================

for _, row in fact_production.iterrows():

    # Probabilidad de requerir mantenimiento
    maintenance_flag = np.random.choice(
        [0,1],
        p=[0.75,0.25]
    )

    if maintenance_flag == 1:

        # ---------------------------------
        # Tipo de mantenimiento
        # ---------------------------------

        maintenance_type = np.random.choice(
            maintenance_types,
            p=maintenance_probs
        )

        planned_flag = (
            maintenance_type != "Corrective"
        )

        # ---------------------------------
        # Orden de trabajo
        # ---------------------------------

        work_order = (
            f"WO{maintenance_counter:07d}"
        )

        # ---------------------------------
        # Horas de mantenimiento
        # ---------------------------------

        maintenance_hours = np.random.uniform(
            1,
            8
        )

        downtime_hours = (
            maintenance_hours
            *
            np.random.uniform(
                0.8,
                1.2
            )
        )

        # ---------------------------------
        # Costos
        # ---------------------------------

        labor_cost = np.random.uniform(
            200_000,
            800_000
        )

        parts_cost = np.random.uniform(
            100_000,
            3_000_000
        )

        maintenance_cost = (
            labor_cost
            +
            parts_cost
        )

        # ---------------------------------
        # MTBF (horas)
        # ---------------------------------

        mtbf = np.random.uniform(
            150,
            600
        )

        # ---------------------------------
        # MTTR (horas)
        # ---------------------------------

        mttr = np.random.uniform(
            1,
            12
        )

        # ---------------------------------
        # Failure Rate λ
        # ---------------------------------

        failure_rate = 1 / mtbf

        # ---------------------------------
        # Tiempo de operación t
        # ---------------------------------

        t = np.random.uniform(
            10,
            mtbf
        )

        # ---------------------------------
        # Reliability
        # R(t)=e^(-t/MTBF)
        # ---------------------------------

        reliability = np.exp(
            -t / mtbf
        )

        # ---------------------------------
        # Maintainability
        # M(t)=1-e^(-t/MTTR)
        # ---------------------------------

        maintainability = (

            1

            -

            np.exp(
                -maintenance_hours / mttr
            )

        )

        # ---------------------------------
        # Maintenance Efficiency
        # ---------------------------------

        maintenance_efficiency = np.random.uniform(
            0.75,
            0.98
        )

        # ---------------------------------
        # Preventive Compliance
        # ---------------------------------

        if maintenance_type == "Preventive":

            preventive_compliance = np.random.uniform(
                0.90,
                1.00
            )

        else:

            preventive_compliance = np.nan

        # ---------------------------------
        # Failure Flag
        # ---------------------------------

        failure_flag = (
            maintenance_type == "Corrective"
        )

        # ---------------------------------
        # Append
        # ---------------------------------

        maintenance_rows.append({

            "MaintenanceID":
            f"MNT{maintenance_counter:08d}",

            "DateKey":
            row["DateKey"],

            "LineID":
            row["LineID"],

            "MaintenanceType":
            maintenance_type,

            "PlannedFlag":
            planned_flag,

            "FailureFlag":
            failure_flag,

            "WorkOrderID":
            work_order,

            "MaintenanceHours":
            round(
                maintenance_hours,
                2
            ),

            "DowntimeHours":
            round(
                downtime_hours,
                2
            ),

            "LaborCost":
            round(
                labor_cost,
                2
            ),

            "PartsCost":
            round(
                parts_cost,
                2
            ),

            "MaintenanceCost":
            round(
                maintenance_cost,
                2
            ),

            "MTBF":
            round(
                mtbf,
                2
            ),

            "MTTR":
            round(
                mttr,
                2
            ),

            "FailureRate":
            round(
                failure_rate,
                6
            ),

            "OperatingTime":
            round(
                t,
                2
            ),

            "Reliability":
            round(
                reliability,
                4
            ),

            "Maintainability":
            round(
                maintainability,
                4
            ),

            "MaintenanceEfficiency":
            round(
                maintenance_efficiency,
                4
            ),

            "PreventiveCompliance":
            preventive_compliance

        })

        maintenance_counter += 1

# ==========================================
# CREATE DATAFRAME
# ==========================================

fact_maintenance = pd.DataFrame(
    maintenance_rows
)

print(
    "Rows:",
    len(fact_maintenance)
)

fact_maintenance.head()
# =========================================
# EXPORTAR CSV
# =========================================

fact_maintenance.to_csv(
    "data/facts/fact_maintenance.csv",
    index=False
)
````
