# 📅 Paso 2: Creación de tabla calendario (dim_fecha)

## 🎯 Objetivo
Construir una tabla calendario que permita realizar análisis temporal en el modelo de datos, como:
- Tendencias de ventas
- Comparaciones Year over Year (YoY)
- Métricas acumuladas (YTD, MTD)

Esta tabla será fundamental para aplicar inteligencia de tiempo en Power BI.

---

## 🧠 Enfoque

Se creó una tabla calendario dinámica utilizando funciones DAX, de modo que el rango de fechas:
- Inicie en la **fecha mínima** de ventas
- Termine en la **fecha máxima** de ventas

Esto garantiza que la tabla se adapte automáticamente a los datos disponibles.

---

## 🛠️ Creación de la tabla en DAX

Se utilizó la función `CALENDAR` para generar el rango de fechas y `ADDCOLUMNS` para enriquecer la tabla con atributos de tiempo.

```DAX id="dimfecha01"
dim_fecha =
ADDCOLUMNS(
    CALENDAR(
        MIN(hecho_ventas_propiedades[fecha_venta]),
        MAX(hecho_ventas_propiedades[fecha_venta])
    ),
    "Año", YEAR([Date]),
    "Mes", FORMAT([Date], "MMMM"),
    "Mes Numero", MONTH([Date]),
    "Año-Mes", FORMAT([Date], "YYYY-MM")
)

# 📊 Paso 4: Creación de medidas

En este paso se desarrollan las medidas analíticas en Power BI (DAX) necesarias para responder las preguntas de negocio del dashboard.

Estas medidas permiten analizar el desempeño general, entender la contribución de distintas dimensiones y evaluar la evolución temporal del negocio.

---

### 4.1 🔹 Medidas base

Estas métricas permiten tener una visión general del desempeño comercial.

- **💰 Ingreso Total**

Ingreso Total = SUM(hecho_ventas_propiedades[precio_venta])

- **📦 Cantidad de Ventas**

Cantidad de Ventas = COUNT(hecho_ventas_propiedades[id_venta])


- **🎯 Ticket Promedio**

Ticket Promedio = DIVIDE([Ingreso Total], [Cantidad de Ventas])


- **💸 Comisión Total**

Comisión Total = SUM(hecho_ventas_propiedades[monto_comision])

---

### 4.2 🔹 Medidas con contexto de filtro

Estas medidas permiten analizar la participación porcentual del ingreso según distintas dimensiones del negocio.

Se utiliza `CALCULATE` junto con `ALL()` para eliminar el filtro de la dimensión en el denominador.

Las medidas que deben construirse son:

- **🏠 Participación de ingresos por tipo de propiedad**

% Ingreso Tipo Propiedad =
DIVIDE(
    [Ingreso Total],
    CALCULATE(
        [Ingreso Total],
        ALL(dim_propiedades[tipo_propiedad])
    )
)

- **📊 Participación de ingresos por canal de venta**

% Ingreso Canal Venta =
DIVIDE(
    [Ingreso Total],
    CALCULATE(
        [Ingreso Total],
        ALL(hecho_ventas_propiedades[canal_venta])
    )
)

- **👤 Participación de ingresos por segmento de cliente**

% Ingreso Segmento Cliente =
DIVIDE(
    [Ingreso Total],
    CALCULATE(
        [Ingreso Total],
        ALL(dim_clientes[segmento_comprador])
    )
)

**💡 Interpretación:**

- El numerador respeta el contexto actual (categoría seleccionada).
- El denominador elimina ese filtro para obtener el total general.
- Estas medidas son ideales para:
    - Tooltips
    - Etiquetas en gráficos
    - Análisis comparativo

---

### 4.3 🔹 Inteligencia de tiempo

Estas medidas permiten analizar la evolución del negocio utilizando la tabla calendario **dim_fecha**.

- **📅 Ventas Year to Date (YTD)**

Ventas YTD =
TOTALYTD(
    [Ingreso Total],
    dim_fecha[Date]
)

- **📆 Ventas Month to Date (MTD)**

Ventas MTD =
TOTALMTD(
    [Ingreso Total],
    dim_fecha[Date]
)

- **⏳ Ventas Año Anterior**

Ventas Año Anterior =
CALCULATE(
    [Ingreso Total],
    SAMEPERIODLASTYEAR(dim_fecha[Date])
)

- **📈 Crecimiento Year over Year (YoY)**

Crecimiento YoY =
DIVIDE(
    [Ingreso Total] - [Ventas Año Anterior],
    [Ventas Año Anterior]
)

**💡 Interpretación:**

- Permiten responder:
  - ¿El negocio está creciendo?
  - ¿Cómo va el desempeño acumulado?
  - ¿Cómo se compara con el año anterior?

---

### 4.4 🔹 Medidas para análisis de cohortes (Columnas calculadas)

Estas columnas permiten analizar la **recurrencia de clientes** y construir la matriz de cohortes.
⚠️ Todas se crean en la tabla `hecho_ventas_propiedades`.

- **🧩 Primera compra por cliente**

Primera Compra =
CALCULATE(
    MIN(hecho_ventas_propiedades[fecha_venta]),
    ALLEXCEPT(
        hecho_ventas_propiedades,
        hecho_ventas_propiedades[id_cliente]
    )
)

- **🗂️ Mes Cohorte**

Mes Cohorte =
FORMAT([Primera Compra], "YYYY-MM")

- **🗓️ Mes Venta**

Mes Venta =
FORMAT(hecho_ventas_propiedades[fecha_venta], "YYYY-MM")


**💡 Interpretación:**

- **Mes Cohorte** → cuándo el cliente fue adquirido.
- **Mes Venta** → cuándo vuelve a comprar.
- Permiten construir una matriz donde:
   - Filas = Cohortes
   - Columnas = Periodos de compra
   - Valores = Ventas o ingresos
