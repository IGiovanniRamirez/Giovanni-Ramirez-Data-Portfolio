```python
# Conversión de fechas

orders['fecha_hora_pedido'] = pd.to_datetime(orders['fecha_hora_pedido'], errors='coerce')
marketing['fecha'] = pd.to_datetime(marketing['fecha'], errors='coerce')
```

```python
# Limpieza dataset Orders

# Estandarizar texto
orders['pais'] = orders['pais'].str.strip().str.lower().str.capitalize()

# Eliminar duplicados por id_pedido
orders = orders.drop_duplicates(subset=['id_pedido'])

# Eliminar valores numéricos inválidos
orders = orders[
    (orders['cantidad'] > 0) &
    (orders['precio_unitario'] > 0) &
    (orders['monto_total'] > 0)
]

# Eliminar outliers extremos
orders = orders[orders['cantidad'] < 1000]

# Validar consistencia de montos
orders['monto_calculado'] = orders['cantidad'] * orders['precio_unitario'] - orders['monto_descuento']

# Mantener registros consistentes (tolerancia pequeña por decimales)
orders = orders[np.isclose(orders['monto_total'], orders['monto_calculado'], atol=1)]

# Eliminar columna auxiliar
orders = orders.drop(columns=['monto_calculado'])

# Manejo de nulos
orders = orders.dropna(subset=[
    'pais', 'dispositivo', 'fuente_referencia',
    'nombre_producto', 'categoria_producto'
])
```

```python
# Limpieza dataset Catalog

# Estandarizar categorías (quitar tilde)
catalog['categoria_producto'] = catalog['categoria_producto'].str.replace('Electrónica', 'Electronica')
```

```python
# Limpieza dataset Marketing

# Completar nulos en canal usando id_campaña
marketing['canal'] = marketing['canal'].fillna(
    marketing['id_campaña'].str.rsplit('_', n=1).str[0]
)

# Estandarizar texto en país
marketing['pais'] = marketing['pais'].str.strip().str.lower().str.capitalize()
```

```python
# Validación final

# Explorar la estructura de orders
display("Orders limpio: vista general de la tabla", orders.head())
print(f"\n")
display("Orders limpio: información general del dataset", orders.info())
print(f"\n")
# Explorar columnas numéricas de orders
display("Orders limpio: estadísticas de columnas numéricas", orders.describe())
print(f"\n")
# Explorar columnas categóricas de orders
categoricas_orders = ['id_pedido', 'id_usuario', 'pais','dispositivo','fuente_referencia','nombre_producto','categoria_producto']
for col in categoricas_orders:
  print(f"\n")
  display("Orders limpio: estadísticas de columnas categóricas",orders[col].value_counts())
  print(f"\n")
# Explorar la estructura de catalog
display("Catalog limpio: vista general de la tabla", catalog.head())
print(f"\n")
display("Catalog limpio: información general del dataset", catalog.info())
print(f"\n")
# Explorar columnas numéricas de catalog
display("Catalog limpio: estadísticas de columnas numéricas", catalog.describe())
print(f"\n")
# Explorar columnas categóricas de catalog
categoricas_catalog = ['nombre_producto', 'categoria_producto', 'proveedor']
for col in categoricas_catalog:
  print(f"\n")
  display("Catalog limpio: estadísticas de columnas categóricas",catalog[col].value_counts())
  print(f"\n")
# Explorar la estructura de marketing
display("Marketing limpio: vista general de la tabla", marketing.head())
print(f"\n")
display("Marketing limpio: información general del dataset", marketing.info())
print(f"\n")
# Explorar columnas numéricas de marketing
display("Marketing limpio: estadísticas de columnas numéricas", marketing.describe())
print(f"\n")
# Explorar columnas categóricas de marketing
categoricas_marketing = ['pais', 'id_campaña', 'canal']
for col in categoricas_marketing:
  print(f"\n")
  display("Marketing limpio: estadísticas de columnas categóricas",marketing[col].value_counts())
```

```python
# 🔹 PASO 2 - ANALIZAR SI EL NEGOCIO ES RENTABLE

# Cargar datasets limpios

orders = pd.read_csv('/content/orders_clean.csv')
catalog = pd.read_csv('/content/catalog_clean.csv')
marketing = pd.read_csv('/content/marketing_clean.csv')

# Merge para traer costo_unitario al dataset de orders
orders_full = orders.merge(
    catalog[['nombre_producto', 'costo_unitario']],
    on='nombre_producto',
    how='left'
)

# ------------------------------
# PARTE 1: RENTABILIDAD
# ------------------------------

# Revenue total
revenue = orders_full['monto_total'].sum()

# Costo total
orders_full['costo_total'] = orders_full['cantidad'] * orders_full['costo_unitario']
costo_total = orders_full['costo_total'].sum()

# Gasto en marketing
marketing_total = marketing['gasto'].sum()

# 2.4 Profit
profit = revenue - costo_total - marketing_total

# ------------------------------
# PARTE 2: COMPORTAMIENTO
# ------------------------------

# Número de órdenes
num_orders = orders_full['id_pedido'].nunique()

# Ticket promedio
ticket_promedio = revenue / num_orders

# Cantidad promedio de productos por orden
cantidad_promedio = orders_full['cantidad'].sum() / num_orders

# Producto más vendido
producto_mas_vendido = (
    orders_full.groupby('nombre_producto')['cantidad']
    .sum()
    .sort_values(ascending=False)
    .idxmax()
)

# Gasto en marketing por canal
gasto_por_canal = marketing.groupby('canal')['gasto'].sum().sort_values(ascending=False)

# ------------------------------
# 3. RESULTADOS
# ------------------------------

print("--- RENTABILIDAD ---")
print(f"Revenue total: {revenue:,.2f}")
print(f"Costo total: {costo_total:,.2f}")
print(f"Gasto en marketing: {marketing_total:,.2f}")
print(f"Profit: {profit:,.2f}")

if profit > 0:
    print("El negocio es rentable")
else:
    print("El negocio NO es rentable")

print("\n --- COMPORTAMIENTO DE VENTAS ---")
print(f"Ticket promedio: {ticket_promedio:,.2f}")
print(f"Cantidad promedio por orden: {cantidad_promedio:,.2f}")
print(f"Producto más vendido: {producto_mas_vendido}")

print("\n --- GASTO POR CANAL ---")
display(gasto_por_canal)
```

  - `begin_checkout` → `add_payment_info`

### 🧠 Insight preliminar

## 📉 Parte 2: Análisis de conversión

### 📊 Tasas de conversión por etapa

| Etapa               | Usuarios | Usuarios previos | Tasa conversión | Drop-off |
|--------------------|----------|------------------|------------------|----------|

### 🔍 ¿Cómo se interpreta la tasa de conversión?

- La tasa de conversión indica qué porcentaje de usuarios avanza al siguiente paso
- El drop-off representa la pérdida de usuarios en cada etapa

👉 Idealmente:
- Conversión alta ✅

```python
# PARTE 1: Totales del funnel
# ======================

query_totals = '''
SELECT
    nombre_evento,
    COUNT(DISTINCT id_usuario) AS usuarios_unicos
FROM events
GROUP BY nombre_evento
ORDER BY usuarios_unicos DESC;
'''

totals = pd.read_sql(query_totals, con=engine)
totals
```

```python
# PARTE 2: Conversiones
# ======================

query_conversion = '''
-- Conteo de usuarios únicos por etapa del funnel y ordenar eventos según el flujo del usuario
WITH funnel AS (
    SELECT
        nombre_evento,
        COUNT(DISTINCT id_usuario) AS usuarios,
        CASE
            WHEN nombre_evento = 'first_visit' THEN 1
            WHEN nombre_evento = 'add_to_cart' THEN 2
            WHEN nombre_evento = 'select_item' THEN 3
            WHEN nombre_evento = 'begin_checkout' THEN 4
            WHEN nombre_evento = 'add_payment_info' THEN 5
            WHEN nombre_evento = 'purchase' THEN 6
        END AS orden
    FROM events
    GROUP BY nombre_evento
),
-- Calcular conversiones entre etapas
conversiones AS (
    SELECT
        nombre_evento,
        usuarios,
        orden,
        LAG(usuarios) OVER (ORDER BY orden) AS usuarios_previos,
        ROUND(usuarios * 100 / LAG(usuarios) OVER (ORDER BY orden), 2) AS tasa_conversion
    FROM funnel
),
--calculo de la tasa de conversion final
conversion_final AS (
    SELECT
        ROUND(
            (SELECT COUNT(DISTINCT id_usuario)
             FROM events
             WHERE nombre_evento = 'purchase') * 100.0
            /
            (SELECT COUNT(DISTINCT id_usuario)
             FROM events
             WHERE nombre_evento = 'first_visit'),
        2) AS conversion_final
)
-- resultado final del funnel
SELECT
    c.nombre_evento,
    c.usuarios,
    c.usuarios_previos,
    c.tasa_conversion,
    ROUND(100 - c.tasa_conversion, 2) AS drop_off,
    cf.conversion_final
FROM conversiones c
CROSS JOIN conversion_final cf
ORDER BY c.orden;
'''

conversion = pd.read_sql(query_conversion, con=engine)
conversion
```

```python
# 🔹 PASO 4 - EVALUAR SI LOS USUARIOS REGRESAN (RETENCIÓN POR COHORTES)
# Explorar tabla users
# =========================
query_users = '''
SELECT *
FROM users
LIMIT 10;
'''
users = pd.read_sql(query_users, con=engine)
users.head(10)
```

```python
# Explorar tabla user_activity
# =========================
query_user_activity = '''
SELECT *
FROM user_activity
LIMIT 10
'''
user_activity = pd.read_sql(query_user_activity, con=engine)
user_activity.head(10)
```

```python
# Retención por cohortes
# ======================

query_cohort_retention_final = '''
-- Convertir fecha de registro a formato DATE
WITH users_clean AS (
  SELECT
  id_usuario,
  CAST(fecha_registro AS DATE) AS fecha_registro
  FROM users
),
-- Definir cohorte por mes de registro
cohortes AS (
    SELECT
        id_usuario,
        TO_CHAR(DATE_TRUNC('month', fecha_registro), 'YYYY-MM') AS cohorte
    FROM users_clean
),
-- Actividad de usuarios con semanas desde registro
actividad AS (
    SELECT
        ua.id_usuario,
        c.cohorte,
        ua.dias_despues_registro,
        CASE
            WHEN ua.dias_despues_registro BETWEEN 1 AND 7 THEN 'w1'
            WHEN ua.dias_despues_registro BETWEEN 8 AND 14 THEN 'w2'
            WHEN ua.dias_despues_registro BETWEEN 15 AND 21 THEN 'w3'
        END AS semana
    FROM user_activity ua
    JOIN cohortes c
        ON ua.id_usuario = c.id_usuario
    WHERE ua.activo = 1
),
-- Usuarios retenidos por cohorte y semana
retencion AS (
    SELECT
        cohorte,
        semana,
        COUNT(DISTINCT id_usuario) AS usuarios_retenidos
    FROM actividad
    WHERE semana IS NOT NULL
    GROUP BY cohorte, semana
),
-- Tamaño inicial de cada cohorte
cohorte_size AS (
    SELECT
        cohorte,
        COUNT(DISTINCT id_usuario) AS usuarios_iniciales
    FROM cohortes
    GROUP BY cohorte
)

-- Resultado final con métricas de retención
SELECT
    r.cohorte,
    cs.usuarios_iniciales,

    -- Retenidos
    SUM(CASE WHEN r.semana = 'w1' THEN r.usuarios_retenidos ELSE 0 END) AS retenido_w1,
    SUM(CASE WHEN r.semana = 'w2' THEN r.usuarios_retenidos ELSE 0 END) AS retenido_w2,
    SUM(CASE WHEN r.semana = 'w3' THEN r.usuarios_retenidos ELSE 0 END) AS retenido_w3,

    -- Porcentajes de retención
    ROUND(SUM(CASE WHEN r.semana = 'w1' THEN r.usuarios_retenidos ELSE 0 END) * 100 / cs.usuarios_iniciales, 2) AS semana_1,
    ROUND(SUM(CASE WHEN r.semana = 'w2' THEN r.usuarios_retenidos ELSE 0 END) * 100 / cs.usuarios_iniciales, 2) AS semana_2,
    ROUND(SUM(CASE WHEN r.semana = 'w3' THEN r.usuarios_retenidos ELSE 0 END) * 100 / cs.usuarios_iniciales, 2) AS semana_3

FROM retencion r
JOIN cohorte_size cs
    ON r.cohorte = cs.cohorte

GROUP BY r.cohorte, cs.usuarios_iniciales
ORDER BY r.cohorte;
'''

# Ejecutar la consulta
cohorte_final = pd.read_sql(query_cohort_retention_final, con=engine)
cohorte_final
```

- **H₁ (Hipótesis alternativa):**  
  La tasa de conversión es diferente en las páginas control y tratamiento. Hay un factor que influye en la decisión de compra.

# ⚙️ Test estadístico aplicado

# 🔍 Interpretación estadística

- El **valor p (0.4161)** es mayor que α (0.05)
- **Interpretación práctica:**
  - Valor p bajo → Hay evidencia estadística contra la hipótesis nula.
  - Valor p alto → No hay evidencia suficiente para afirmar una diferencia en conversión

👉 Esto significa que:

❌ **No se rechaza la hipótesis nula**

- No hay evidencia estadísticamente significativa  
- La diferencia observada puede explicarse por el azar

```python
# Importar libreria statsmodels
from statsmodels.stats.proportion import proportions_ztest

# Número de usuarios convertidos por página
conversiones = checkout.groupby('variante')['convirtio'].sum()

# Total de usuarios por página
totales = checkout.groupby('variante')['convirtio'].count()

display("Usuarios convertidos por página:", conversiones)
print(f"\n")
display("Total de usuarios por página:", totales)
```

```python
# Aplicar prueba
exitos = [conversiones['control'], conversiones['tratamiento']]
observaciones = [totales['control'], totales['tratamiento']]
z_stat, p_value = proportions_ztest(exitos , observaciones)
tasa_control = exitos[0] / observaciones[0]
tasa_tratamiento = exitos[1] / observaciones[1]
# Visualizar resultados
print(f"Estadístico : {z_stat}")
print(f"Valor p: {p_value}")
alpha = 0.05  # umbral de significancia

if p_value < alpha:
    print("Rechazamos la hipótesis nula: hay evidencia de una diferencia.")
else:
    print("No rechazamos la hipótesis nula: no hay evidencia suficiente de una diferencia.")

print(f"\nTasa de conversión página Control: {tasa_control:.2%}")
print(f"Tasa de conversión página Tratamiento: {tasa_tratamiento:.2%}")
```
